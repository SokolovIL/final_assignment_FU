# Шаг 10. Проверка пропусков

Будет проведён подсчёт пустых ячеек, анализ причин их появления и выбор стратегию обработки.

```python
import pandas as pd

# Загружаем консолидированный датасет (длинный формат)
df = pd.read_csv('df_consolidated.csv')

print("=== 1. Абсолютное количество пропусков ===")
print(df.isna().sum())

print("\n=== 2. Доля пропусков в процентах ===")
print((df.isna().mean() * 100).round(2).astype(str) + '%')
```

## Анализ причин появления пропусков
Основываясь на природе данных (обзоры пластиковых миниатюр), пропуски не являются ошибкой, они несут аналитический смысл.
- release_year (~15% пропусков): Год выпуска не указан на сайте для наборов 1950-60-х годов или для наборов, которые являются прямыми переизданиями (reissue) без новой даты;
- aggregate_rating (~10-15% пропусков): Рейтинги отсутствуют для наборов аксессуаров, техники (без фигур) или наборов, которые на сайте упомянуты, но не оценены.
- num_figures (~10% пропусков): Количество фигур равно NaN для наборов, состоящих исключительно из техники, зданий или животных.
- nationality (~9% пропусков): Национальность не указана для универсальных наборов.

## Стратегия работы с пропусками
Строки с пропусками удаляться не будут, так как удаление строк исказит общую картину ассортимента производителей.

## Результат шага
Выявлены пропуски и рассчитана их доля.
Установлено, что пропуски в числовых метриках (release_year, rating, figures) являются информативными (отражают специфику наборов-аксессуаров или старых релизов).
Выбрана стратегия: числовые поля будут оставлены как NaN (с типом Int64), а текстовое поле nationality будет заполнено значением 'Не указано' для корректной работы будущих сводных таблиц и BI-фильтров.
