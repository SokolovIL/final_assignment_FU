# Шаг 3. Ожидаемый результат анализа
Итоговый результат состоит из четырёх ключевых блоков.
## 1. Файл с очищенными и обогащёнными данными
### Формат:
CSV-файл.
### Содержание:
Единая консолидированная таблица в длинном формате, готовая для загрузки в BI-систему.

**Единица наблюдения:** одна связка «Набор — Национальность».

**Базовые поля:** ID, Header, Manufacturer, Years_from, Years_to, Nationality, Release_Year, Aggregate_Rating, Num_Figures.

**Расчётные признаки:**
- Period_Duration — длительность исторического периода.
- Mid_Year — середина периода.
- Era — историческая эра (Античность, Средние века, Новое время, Новейшее время).
- Decade — десятилетие выпуска набора.
- Years_Since_Release — сколько лет прошло с момента выпуска.
## 2. Таблицы с агрегированными показателями и сегментация объектов
### Формат:
Сводные таблицы и витрины данных.
### Содержание:
- ABC-сегментация: разбивка производителей и исторических эпох на категории A (ядро рынка, 80% наборов), B (середина), C (нишевые темы).
- XYZ-сегментация: стабильность релизов производителей (X — стабильные, Y — нерегулярные, Z — спорадические).
- Адаптированный RFM-анализ: для каждой связки «Производитель — Национальность» рассчитаны метрики Recency (давность последнего релиза), Frequency (общее количество наборов), Magnitude (средний рейтинг качества).
## 3. Интерактивный дашборд и графики
### Формат:
Дашборд в BI-платформе с оформленной аналитической историей.
### Структура экранов (с учётом приоритетов):
**Экран 1 «Карта эпох и национальностей»** (наивысший приоритет) - Тепловая карта или древовидная карта
- Показывает, какие национальности в какие исторические периоды представлены лучше всего, а какие хуже всего.
- Позволяет коллекционеру увидеть покрытие интересующей его ниши.
  
**Экран 2 «Обзор рынка»**
- Динамика выпусков наборов по годам и десятилетиям.
- Топ производителей по количеству наборов.
  
**Экран 3 «Качество и Эволюция»**
- Связь количества фигур и рейтинга качества.
- Распределение рейтингов по производителям.
- Динамика среднего рейтинга по годам выпуска.
  
**Экран 4 «Гид коллекционера»** (второй по приоритету) - Интерактивные фильтры по эпохе и национальности
- Предоставляет список доступных производителей, отсортированный по рейтингу и актуальности.
- Позволяет пользователю выбрать конкретную нишу и увидеть лучшие наборы для неё.
## 4. Аналитическая записка с инсайтами и рекомендациями
### Формат:
Текстовый документ.
### Содержание:
**Факты из данных:** краткая сводка по структуре рынка.

**Выявленные инсайты:** ключевые паттерны активности производителей, эволюция качества, слабые и перспективные ниши.

**Ограничения анализа:** указание на отсутствие данных о реальных продажах и вторичном рынке, субъективность рейтингов.

**Практические рекомендации:** готовые советы для коллекционеров.
