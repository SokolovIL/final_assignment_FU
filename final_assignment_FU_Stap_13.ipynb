# Шаг 13. Проверка числовых столбцов

Будет осуществена проверка данных на наличие логических противоречий или невозможных значений, которые могли возникнуть из-за ошибок парсинга или ввода.

## Объекты проверки:
- years_from > years_to: Исторический период не может идти в обратном направлении.
- num_figures < 0 или > 200: Отрицательное количество фигур невозможно, а наборы более 200 фигур встречаются крайне редко (обычно это диорамы или сборные комплекты, но значение >200 стоит проверить на ошибку парсинга).
- aggregate_rating < 5 или > 50: Поскольку рейтинг складывается из 5 оценок по 10-балльной шкале, минимально возможная сумма равна 5, а максимальная — 50.
- release_year < 1950 или > 2026: Массовое производство пластиковых солдатиков началось в 1950-х годах. Значения раньше 1950 года или в будущем (позже 2026) требуют ручной проверки.
- period_duration < 0: Длительность исторического периода не может быть отрицательной.

```python
import pandas as pd
import numpy as np

# Загружаем очищенный датасет с правильными типами данных
df = pd.read_csv('df_consolidated.csv', dtype={
    'release_year': 'Int64',
    'aggregate_rating': 'Int64',
    'num_figures': 'Int64',
    'decade': 'Int64',
    'years_since_release': 'Int64'
})

print("=== 1. Проверка логических противоречий в исторических периодах ===")
# years_from не может быть больше years_to
invalid_periods = df[df['years_from'] > df['years_to']]
print(f"Наборов, где год начала больше года окончания: {len(invalid_periods)}")
if len(invalid_periods) > 0:
    display(invalid_periods[['id', 'header', 'years_from', 'years_to']].head())

print("\n=== 2. Проверка аномалий в количестве фигур (num_figures) ===")
# Количество фигур не может быть отрицательным или нереалистично огромным (например, > 200)
invalid_figures = df[(df['num_figures'] < 0) | (df['num_figures'] > 200)]
print(f"Наборов с аномальным количеством фигур: {len(invalid_figures)}")
if len(invalid_figures) > 0:
    display(invalid_figures[['id', 'header', 'num_figures']].head())

print("\n=== 3. Проверка аномалий в рейтинге (aggregate_rating) ===")
# Рейтинг является суммой 5 оценок от 1 до 10, поэтому он должен быть в диапазоне от 5 до 50
invalid_ratings = df[(df['aggregate_rating'] < 5) | (df['aggregate_rating'] > 50)]
print(f"Наборов с аномальным рейтингом: {len(invalid_ratings)}")
if len(invalid_ratings) > 0:
    display(invalid_ratings[['id', 'header', 'aggregate_rating']].head())

print("\n=== 4. Проверка аномалий в годе выпуска (release_year) ===")
# Пластиковые миниатюры в массовом производстве появились не ранее 1950-х годов. 
# Проверим значения за пределами разумного диапазона (например, < 1950 или > 2026)
invalid_years = df[(df['release_year'] < 1950) | (df['release_year'] > 2026)]
print(f"Наборов с аномальным годом выпуска: {len(invalid_years)}")
if len(invalid_years) > 0:
    display(invalid_years[['id', 'header', 'release_year']].head())

print("\n=== 5. Проверка отрицательной длительности периода (period_duration) ===")
invalid_duration = df[df['period_duration'] < 0]
print(f"Наборов с отрицательной длительностью периода: {len(invalid_duration)}")

print("\n=== Итоговый вывод ===")
total_anomalies = len(invalid_periods) + len(invalid_figures) + len(invalid_ratings) + len(invalid_years) + len(invalid_duration)
if total_anomalies == 0:
    print("✅ Аномалий и критических выбросов в данных не обнаружено. Датасет готов к анализу!")
else:
    print(f"⚠️ Обнаружено {total_anomalies} потенциальных аномалий. Рекомендуется проверить их вручную.")
```
