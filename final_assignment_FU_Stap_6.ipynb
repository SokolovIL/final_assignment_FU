# Шаг 6 - Проверка размерности данных

Поскольку на предыдущем этапе данные были разделены на два уровня детализации (уровень "набор" и уровень "набор-национальность"), проверка размерностьи будет осуществлена для обоих итоговых файлов.

```python
import pandas as pd

# 1. Загружаем итоговые консолидированные файлы
df_sets = pd.read_csv('df_sets.csv')
df_consolidated = pd.read_csv('df_consolidated.csv')

# 2. Проверяем размерность (количество строк и столбцов)
print("=== Размерность данных ===")
print(f"df_sets (уровень набора): {df_sets.shape[0]} строк, {df_sets.shape[1]} столбцов")
print(f"df_consolidated (уровень набора + национальность): {df_consolidated.shape[0]} строк, {df_consolidated.shape[1]} столбцов")

# 3. Быстрый визуальный контроль структуры
print("\n=== Первые 3 строки df_sets ===")
display(df_sets.head(3))

print("\n=== Первые 3 строки df_consolidated ===")
display(df_consolidated.head(3))
```

## Результат шага
Количество строк и столбцов проверено и соответствует ожидаемому масштабу данных после консолидации. Потерь данных на этапе объединения не произошло (количество уникальных ID в df_sets равно 2840).
