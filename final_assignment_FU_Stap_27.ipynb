# Шаг 27. Итоговый аналитический отчёт

## Анализ рынка пластиковых исторических миниатюр (1950-е — 2026 гг.)

**Исполнитель:** Соколов И.Л.  
**Дата подготовки:** август 2026 г.  
**Объект исследования:** 2 840 наборов пластиковых исторических миниатюр от 61 производителя

---

## Резюме для руководителя (Executive Summary)

Рынок пластиковых исторических миниатюр — зрелый, высококонцентрированный и технологически стабильный. Всего **18 производителей (29,5% брендов) формируют 78,9% всего ассортимента**. Среднее качество наборов стабилизировалось на уровне 41–42 балла из 50 начиная с 2000-х годов. Ключевая проблема коллекционера — не отсутствие наборов, а **навигация в избыточном ассортименте**: 81% наборов являются условно избыточными, так как их историческая ниша уже закрыта другими релизами.

Разработанные алгоритмы (адаптированные RFM и XYZ, скоринг редкости с учётом перекрытия периодов) позволяют коллекционеру экономить бюджет, избегая дубликатов, и фокусироваться на стабильных производителях и перспективных нишах.

---

## 1. Цель анализа

Помочь коллекционерам и варгеймерам принимать обоснованные решения при формировании коллекций:
- выявлять редкие ниши и избегать покупки дубликатов;
- выбирать надёжных производителей по конкретным историческим темам;
- отслеживать перспективы развития тем и качество продукции.

**Целевая аудитория:** коллекционеры-любители, варгеймеры, исторические клубы, авторы контента.

---

## 2. Использованные данные

Анализ основан на трёх консолидированных датасетах:

| Источник | Содержание | Единица наблюдения |
|---|---|---|
| `review_periods.csv` | Исторические периоды, охватываемые наборами (от античности до современности) | 1 набор |
| `reviews_by_nationality_wide.csv` | Национальности и армии, представленные в наборах (до 11 на набор) | 1 набор × национальность |
| `psr_enriched_data.csv` | Метаданные: год выпуска, совокупный рейтинг качества (из 50 баллов), количество фигур | 1 набор |

**Итоговый объём:** 2 840 уникальных наборов, 3 937 пар «набор–национальность», 61 производитель, охват периодов от −2024 г. до н.э. до 2024 г.

---

## 3. Краткое описание подготовки данных

Подготовка выполнена в соответствии с методологией CRISP-DM:

1. **Консолидация:** три таблицы объединены по ключу `ID` в единый датасет с двумя уровнями детализации (уровень набора и уровень «набор–национальность»).
2. **Очистка:** удалены полные дубликаты и некорректные записи; объём сократился с 3 937 до 3 506 строк на уровне «набор–национальность».
3. **Обработка пропусков:**
   - категориальные пропуски (`nationality`) заменены на явное значение «Не указано» — это сохранило строки для BI-фильтров;
   - числовые пропуски (`release_year`, `aggregate_rating`, `num_figures`) намеренно оставлены как `NaN` — их заполнение медианой исказило бы историческую реальность.
4. **Обогащение:** рассчитаны производные признаки — длительность исторического периода, середина периода, десятилетие выпуска, давность релиза.
5. **Валидация:** проведена программная проверка выбросов, корреляций и корректности всех агрегаций (9 контрольных тестов пройдены успешно).

---

## 4. Основные расчёты и методы

### 4.1. ABC-анализ производителей
Сегментация 61 бренда по доле в ассортименте:
- **Категория A** (18 брендов) — 78,9% ассортимента;
- **Категория B** (18 брендов) — 16,0%;
- **Категория C** (25 брендов) — 5,2%.

### 4.2. Адаптированный RFM-анализ
Для каждой пары «производитель × национальность» рассчитаны:
- **Recency** — лет с последнего релиза;
- **Frequency** — количество уникальных наборов;
- **Magnitude** — средний рейтинг качества.

Выявлено **74 «актуальных хита»** (Recency ≤ 5 лет, Magnitude ≥ 40).

### 4.3. Адаптированный XYZ-анализ
Оценивается **стабильность присутствия производителя в нише** (не спрос, а регулярность релизов):
- **X** — стабильные релизы из года в год (122 пары);
- **Y** — волнообразные выпуски (109 пар);
- **Z** — эпизодические релизы (14 пар).

### 4.4. Алгоритм выявления избыточности
Для каждой ниши определяется, перекрывается ли исторический период набора периодом другого набора той же национальности. Это позволяет исключить дубликаты при закрытии «белых пятен».

---

## 5. Ключевые визуализации

В проекте построено 10 целевых графиков, каждый из которых отвечает на конкретный аналитический вопрос:

1. **ABC-анализ** — столбчатая диаграмма Топ-40 брендов с цветовой кодировкой категорий.
2. **Динамика рынка** — график с двойной осью (количество наборов и средний рейтинг по десятилетиям).
3. **Размер набора и рейтинг** — сравнение среднего качества по категориям размера.
4. **Тепловая карта покрытия (1939–1945)** — срез плотности релизов по годам и национальностям.
5. **RFM-анализ** — пузырьковая диаграмма актуальных хитов.
6. **Карта исторических периодов** — Топ-5 национальностей по каждой эпохе в хронологическом порядке.
7. **XYZ-анализ** — распределение пар по категориям стабильности.
8. **Лидеры по десятилетиям** — Топ-1 производитель в каждом десятилетии.
9. **Рост тем (2020–2026 vs 2015–2019)** — Топ-10 тем с наибольшим приростом.
10. **Падение тем** — Топ-10 тем с наибольшим снижением активности.

Все графики доступны в файле `final_assignment_FU_Stap_19.ipynb` и подготовлены для переноса в BI-дашборд.

---

## 6. Основные выводы

### Структура рынка
- Рынок высококонцентрирован: 18 брендов формируют 4/5 ассортимента.
- Лидеры менялись по десятилетиям: Airfix (1960-е) → Atlantic (1970-е) → Esci (1980-е) → Revell (1990-е) → HaT (2000-е) → Strelets (2010–2020-е).

### Качество и эволюция
- Средний рейтинг стабилизировался на уровне 41–42 балла с 2000-х годов.
- Крупные наборы (51+ фигур) оцениваются выше (42,3), чем мини-наборы (40,5).

### Динамика тем
- **Рост (2020–2026):** Italian/Древний мир (+129%), USA/Новейшее время (+116%), Italian/Новейшее время (+100%).
- **Падение:** British/Средневековье (−88%), Confederate/Новое время (−85%), Italian/Средневековье (−80%).

### Редкие ниши
- Из 2 026 потенциально редких наборов **1 645 (81%) являются условно избыточными** — их ниша уже закрыта другими релизами.
- Уникально полезными для покупки остаются **381 набор**.

---

## 7. Ограничения анализа

1. **Отсутствие данных о продажах и ценах.** Анализ опирается только на факты релизов и экспертные рейтинги. Реальная популярность (тиражи, выручка) не оценивалась.
2. **Пропуски в ключевых метриках.** У 527 наборов отсутствует год выпуска, у 901 — рейтинг. Временнáя динамика построена на подвыборке.
3. **Смещение источника.** Рейтинги собраны с одного сайта-агрегатора.
4. **Адаптация XYZ-анализа.** Метод оценивает стабильность *предложения*, а не *спроса*. Категория «Z» означает эпизодические релизы, что может быть осознанной стратегией нишевого бренда.
5. **Упрощённая логика избыточности.** Алгоритм не учитывает качество скульптинга и историческую достоверность конкретной модели.
6. **Корреляция ≠ причинность.** Связь между размером набора и рейтингом не доказывает, что увеличение фигур *улучшает* качество — действует скрытый фактор.

## 8. Практические рекомендации

### Для начинающих коллекционеров
Ориентируйтесь на производителей **Категории A** — они формируют 78,9% ассортимента при стабильном качестве 41,8 из 50.

### Для оптимизации бюджета
Перед покупкой «редкого» набора проверьте файл `final_recommendations_rare_niches.csv` с фильтром `is_redundant == False`. Это исключит 81% дубликатов.

### Для варгеймеров (сборка армий 40–100+ фигур)
Выбирайте производителей с категорией стабильности **X** или **Y** в файле `agg_manufacturers_xyz.csv` — это гарантирует регулярные переиздания и совместимые дополнения.

### Для инвесторов-коллекционеров и клубов
Фокусируйтесь на темах с темпом роста >80% за 2020–2026 гг. (файл `promising_themes_growth.csv`) — производители активно перераспределяют ресурсы в эти ниши.

Перед покупкой «редкого» набора проверьте файл `final_recommendations_rare_niches.csv` с фильтром `is_redundant == False`. Это исключит 81% дубликатов.

### Для инвесторов-коллекционеров и клубов
