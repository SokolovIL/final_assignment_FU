# Шаг 1. Цель анализа
## Предмет анализа:
Ландшафт рынка пластиковых исторических миниатюр с 1950-х годов по настоящее время на основе трёх консолидированных датасетов:
1. Исторические периоды (review_periods.csv) — какие эпохи охватывают наборы (от античности до современности)
2. Национальности (reviews_by_nationality_wide.csv) — какие народы и армии представлены в наборах
3. Метаданные наборов (psr_enriched_data.csv) — год выпуска, совокупный рейтинг качества (из 50 возможных баллов), количество фигур в наборе
## Единица наблюдения:
один набор пластиковых исторических миниатюр от конкретного производителя.
## Ключевые поля для анализа:
1. Производитель (извлекается из Header)
2. Исторический период (Years_from, Years_to)
3. Национальности (до 11 на набор)
4. Год выпуска набора (Release_Year)
5. Рейтинг качества (Aggregate_Rating, сумма 5 оценок по 10 баллов)
6. Количество фигур (Num_Figures)
## Цель исследования:
Помочь коллекционерам и варгеймерам принимать обоснованные решения при формировании коллекций.
## Задачи исследования:
1. Выявить слабые ниши — исторические темы, которые производители давно не поддерживают новыми наборами (последний релиз более 15 лет назад). Такие наборы стали редкостью для коллекционеров.
2. Определить перспективные направления — темы, которые активно развиваются производителями в последние 5 лет (много новых качественных наборов).
Найти оптимальных производителей — для каждой исторической эпохи выявить бренды с лучшим соотношением качества и доступности (свежие релизы).
3. Оценить эволюцию качества — как менялся средний рейтинг наборов по годам выпуска и по производителям.
4. Создать систему рекомендаций — показать всех производителей, выпускавших наборы по конкретной теме, отсортированные по рейтингу и актуальности.
## Целевая аудитория исследования:
1. Коллекционеры-любители — люди, собирающие коллекции миниатюр по конкретным историческим периодам.
2. Варгеймеры — игроки в настольные варгеймы, которым нужны большие армии миниатюр (40-100+ фигур)
3. Исторические клубы и сообщества — для организации совместных закупок и обмена опытом.
4. Авторы контента — блогеры, пишущие обзоры на миниатюры.
