# Шаг 8. Изучение типов данных

Получение сводки о структуре датафрейма: имена столбцов, количество непустых значений и типы данных.

```python
import pandas as pd

# 1. Загружаем основной консолидированный датасет (длинный формат)
df = pd.read_csv('df_consolidated.csv')

# 2. Изучаем общую информацию о датасете
print("=== Общая информация о датасете (df.info()) ===")
df.info()

# 3. Дополнительный просмотр типов данных в виде списка для удобства
print("\n=== Типы данных по столбцам ===")
print(df.dtypes)

# 4. Быстрая проверка уникальных значений в категориальных столбцах (опционально, но полезно)
print("\n=== Уникальные значения в столбце 'Era' ===")
print(df['Era'].dropna().unique())
```

Числовые столбцы с пропусками (NaN) определены как float64 (числа с плавающей точкой). Из-за этого в данных и на будущих графиках годы и рейтинги будут отображаться с десятичной частью.

Необходимо привести эти столбцы к типу Int64. 

```python
# Список столбцов, которые должны быть целыми числами, но из-за пропусков стали float64
cols_to_fix = ['Release_Year', 'Aggregate_Rating', 'Num_Figures', 'Decade', 'Years_Since_Release']

# Преобразуем типы данных
for col in cols_to_fix:
    df[col] = df[col].astype('Int64')  # Обратите внимание на заглавную 'I'

# Проверяем результат
print("\n=== Типы данных после исправления ===")
print(df.dtypes)
```

## Результат шага
Типы данных определены. 
Выявлено и устранено присвоение типа float64 целочисленным столбцам, содержащим пропуски (NaN). Данные приведены к корректным типам (Int64 для числовых метрик с пропусками, object для текста), что гарантирует корректную работу математических функций и методов визуализации на следующих этапах.
