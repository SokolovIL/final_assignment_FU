# Шаг 2. Аналитические вопросы

На основе сформулированной цели анализа составлен список конкретных, проверяемых по данным вопросов. Каждый вопрос разбит по тематическим блокам для удобства последующей работы.
## Блок 1: Структура рынка (ABC-анализ)
1. Какие производители выпускают наибольшее количество наборов? Какие 20% производителей дают 80% всего ассортимента?
2. Какие исторические эпохи наиболее представлены на рынке? На какие периоды приходится основная доля всех выпущенных наборов?
3. Какие национальности чаще всего встречаются в наборах? Есть ли национальности, представленные в менее чем 5 наборах (нишевые интересы)?
## Блок 2: Динамика рынка во времени
4. Как менялось количество выпускаемых наборов по годам? Есть ли периоды роста или спада активности производителей?
5. Какие производители были активны в разные десятилетия? Кто доминировал в 1960-х, 1980-х, 2000-х, 2010-х, 2020-х?
6. Есть ли слабые ниши? Какие исторические темы (национальности или периоды) не имели новых выпусков более 15 лет?
7. Какие темы являются перспективными? По каким национальностям или периодам наблюдается рост числа новых наборов за последние 5 лет (2020–2026)?
## Блок 3: Качество наборов (Aggregate_Rating)
8. Как менялся средний рейтинг качества наборов по годам выпуска? Становятся ли современные наборы лучше или хуже старых?
9. Какие производители имеют наивысший средний рейтинг качества? Кто стабильно выпускает высококачественные наборы?
10. Есть ли связь между количеством фигур в наборе и рейтингом качества? Большие наборы (50+ фигур) оцениваются выше или ниже малых (12–24 фигуры)?
## Блок 4: Рекомендации для коллекционеров
11. Для конкретных исторических периодов — какие производители предлагают наборы с лучшим рейтингом? 
12. Есть ли связь между годом выпуска и количеством фигур в наборе? Становятся ли современные наборы больше или меньше по количеству фигур?
## Блок 5: Адаптация RFM и XYZ под контекст коллекционирования
13. RFM-анализ активности производителей: Для каждой исторической темы (национальность или период) — как давно производитель выпускал последний набор (Recency), сколько всего наборов по теме он выпустил (Frequency), и каков средний рейтинг качества (Magnitude вместо Monetary)?
14. XYZ-анализ стабильности релизов: Какие производители выпускают наборы стабильно каждые 1–2 года (категория X), а какие — спорадически, раз в 10+ лет (категория Z)?
