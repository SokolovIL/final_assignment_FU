# Шаг 9. Проверка и стандартизация названий столбцов

```python
import pandas as pd

# --- 1. Загрузка консолидированных данных ---
df_sets = pd.read_csv('df_sets.csv')
df_consolidated = pd.read_csv('df_consolidated.csv')

print("=== Исходные названия столбцов ===")
print("\ndf_sets:")
print(df_sets.columns.tolist())
print("\ndf_consolidated:")
print(df_consolidated.columns.tolist())

# --- 2. Стандартизация названий к формату snake_case ---
# Последовательность операций:
# .str.strip()            — убирает случайные пробелы в начале/конце
# .str.lower()            — переводит в нижний регистр
# .str.replace(' ', '_')  — заменяет пробелы на подчеркивания
df_sets.columns = df_sets.columns.str.strip().str.lower().str.replace(' ', '_')
df_consolidated.columns = df_consolidated.columns.str.strip().str.lower().str.replace(' ', '_')

# --- 3. Проверка результата ---
print("\n=== Названия столбцов после стандартизации ===")
print("\ndf_sets:")
print(df_sets.columns.tolist())
print("\ndf_consolidated:")
print(df_consolidated.columns.tolist())

# --- 4. Визуальный контроль (первые 3 строки с новыми названиями) ---
print("\n=== Пример данных df_consolidated (первые 3 строки) ===")
display(df_consolidated.head(3))

# --- 5. Проверка на дубликаты названий (важно для корректности join-операций) ---
print("\n=== Проверка на дубликаты названий столбцов ===")
print(f"df_sets: дубликатов нет — {not df_sets.columns.duplicated().any()}")
print(f"df_consolidated: дубликатов нет — {not df_consolidated.columns.duplicated().any()}")

# --- 6. Сохранение результатов в CSV ---
df_sets.to_csv('df_sets.csv', index=False, encoding='utf-8')
df_consolidated.to_csv('df_consolidated.csv', index=False, encoding='utf-8')

print("\n=== Файлы обновлены и сохранены ===")
print(f"df_sets: {df_sets.shape[0]} строк, {df_sets.shape[1]} столбцов")
print(f"df_consolidated: {df_consolidated.shape[0]} строк, {df_consolidated.shape[1]} столбцов")
```

## Результат 
Названия столбцов проверены, очищены от потенциальных артефактов и приведены к единому и стандартному для анализа формату (snake_case).
