# Шаг 25. Ограничения анализа

Систематизированы ключевые ограничения, выявленные в ходе работы с датасетами пластиковых исторических миниатюр. 

Ограничения разделены на три категории: ограничения данных, методологические ограничения и ограничения интерпретации.

## 1. Ограничения, связанные с качеством и полнотой данных

- **Отсутствие данных о продажах и ценах:** В датасетах присутствуют только факты релизов и экспертные рейтинги качества. Отсутствуют данные о тиражах, объеме продаж и розничной стоимости наборов. Это делает невозможным анализ реальной популярности (спроса) или соотношения "цена/качество".
- **Пропуски в ключевых метриках:** У значительной части наборов отсутствуют значения `release_year` (527 строк) и `aggregate_rating` (901 строка). Несмотря на то, что пропуски корректно обработаны как `NaN` (с типом `Int64`), это означает, что временнáя динамика и анализ качества построены только на подвыборке данных, а не на всей генеральной совокупности (2840 наборов).
- **Смещение источника (Source Bias):** Данные о рейтингах собраны с одного конкретного сайта-агрегатора обзоров. Оценки могут быть субъективны, а для нишевых производителей рейтинг может быть основан всего на небольшом количестве оценок, что статистически ненадёжно.
- **Неполнота недавних периодов:** Данные за 2024–2026 годы могут быть неполными, так как новые наборы часто появляются в продаже раньше, чем на них появляются подробные обзоры и оценки на профильных ресурсах.

## 2. Методологические ограничения

- **Адаптация XYZ-анализа:** Классический XYZ анализирует стабильность *спроса*. Ввиду отсутствия данных о продажах, метод был адаптирован для оценки стабильности *предложения* (regularity релизов). Поэтому категория "Z" означает не "низкий спрос", а "эпизодические выпуски", что для нишевых коллекционных брендов может быть осознанной бизнес-стратегией, а не признаком проблем.
- **Упрощенная логика избыточности (is_redundant):** Алгоритм помечает набор как избыточный, если его исторический период математически полностью вкладывается в период другого набора той же национальности. Этот алгоритм не учитывает качество скульптинга, историческую достоверность фигур или разнообразие поз, которые могут сделать "избыточный" набор коллекционно более ценным.
- **Агрегация по крупным периодам:** Группировка по десятилетиям или широким историческим периодам сглаживает важные краткосрочные тренды, связанные с конкретными историческими событиями.

## 3. Ограничения интерпретации (Риск ложных причинно-следственных связей)

- **Корреляция размера и рейтинга:** Выявлена слабая положительная связь между количеством фигур (`num_figures`) и рейтингом (`aggregate_rating`). Нельзя утверждать, что увеличение числа фигур *причинно улучшает* качество. 
- **Объединение категорий:** При обработке пропусков категориальное значение `nationality` было заменено на "Не указано". Хотя это позволило сохранить строки в датасете, эти наборы были исключены из анализа ниш, что сужает охват исследования.

## Влияние ограничений на итоговые выводы

Несмотря на указанные ограничения, проведенный анализ остается валидным и полезным для целевой аудитории (коллекционеров и варгеймеров), при условии соблюдения следующих правил интерпретации:

1. **Рекомендации по качеству** следует рассматривать как экспертную оценку конкретного источника, а не как объективную истину.
2. **RFM и XYZ-метрики** работают для фильтрации "информационного шума" и выявления стабильных брендов, но не должны быть единственным критерием при поиске уникальных, редких наборов категории "Z".
3. **Анализ редких ниш** (с фильтрацией `is_redundant == False`) дает стартовый чек-лист, но финальное решение о покупке должно учитывать коллекционную ценность конкретного набора.
