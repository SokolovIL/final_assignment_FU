# Шаг 4. Источники данных
Единый источник данных
## Название источника
Сайт Plastic Soldier Review (https://www.plasticsoldierreview.com)
## Тип источника
Веб-сайт (независимый ресурс с обзорами наборов военно-исторических миниатюр, работает с 2002 года).
## Способ получения данных
Данные извлечены автоматически путём эмуляции HTTP-запросов к страницам обзоров сайта plasticsoldierreview.com. Логика сбора, скрипты парсинга и промежуточные преобразования зафиксированы в файле final_assignment_FU_Stap_0.ipynb, размещённом в репозитории.
Дата выгрузки: 17 августа 2026 года.
## Владелец данных
Пользователь SokolovIL — владелец репозитория SokolovIL/final_assignment_FU, в котором хранятся результаты парсинга.
## Результат парсинга
Три датасета.
Все три CSV-файла являются производными продуктами одного и того же источника и созданы в ходе работы пользователя.

**1. review_periods.csv — Исторические периоды наборов**

**- Структура:** 2837 строк, 4 колонки (ID, Header, Years_from, Years_to)

**- Единица наблюдения:** один набор миниатюр

**- Содержание:** идентификатор обзора, название набора (включая производителя и артикул), границы исторического периода, который охватывает набор (включая отрицательные значения для годов до н.э.)

**- Способ извлечения:** парсинг заголовков обзоров и описательных разделов сайта

**2. reviews_by_nationality_wide.csv — Национальности в наборах**

**- Структура:** 2476 строк, 13 колонок (ID, Header, Nationality_1 ... Nationality_11)

**- Единица наблюдения:** один набор миниатюр

**- Содержание:** идентификатор обзора, название набора, до 11 национальностей миниатюр в наборе (в широком формате)

**- Способ извлечения:** парсинг атрибутивных данных из обзоров

**- Особенность:** содержит на 361 строку меньше, чем review_periods.csv — для этих наборов информация о национальностях отсутствует на сайте.

**3. psr_enriched_data.csv — Метаданные наборов с сайта**

**- Структура:** 2837 строк, 5 колонок (ID, Header, Release_Year, Aggregate_Rating, Num_Figures). Типы числовых колонок — целые, пропуски представлены как NaN.

**- Единица наблюдения:** один набор миниатюр

**- Содержание:**

    - Release_Year — год выпуска набора (найден для 2406 наборов)
    
    - Aggregate_Rating — сумма пяти экспертных оценок качества (Historical Accuracy, Pose Quality, Pose Number, Sculpting, Mould) по 10-балльной шкале;
    
    - Num_Figures — количество фигур в наборе (найдено для 2490 наборов)

**Способ извлечения:** парсинг таблиц Stats и секций Ratings из страниц обзоров

## Ограничения использования
**1. Субъективность рейтингов**. Aggregate_Rating — это экспертные оценки рецензентов сайта, а не стандартизированная метрика качества. Интерпретировать рейтинги следует с учётом возможной субъективности и смещений рецензентов.

**2. Неравномерность покрытия**. Для части наборов (около 15%) отдельные поля отсутствуют.

**3. Правовой статус**. Данные собраны с публично доступного сайта для целей учебного проекта. Коммерческое использование или массовая публикация требует согласования с владельцем сайта Plastic Soldier Review. В рамках учебного проекта использование допустимо с обязательной атрибуцией источника.

**4. Возможные ошибки парсинга**. Автоматическое извлечение данных из HTML-страниц может содержать единичные ошибки, связанные с нестандартной разметкой отдельных обзоров. Требуется проверка качества данных на Шаге 5.
## Уровень доверия к данным
**Высокий** — для структурных данных. Идентификаторы, названия наборов, годы выпуска, количество фигур и исторические периоды извлекаются из хорошо структурированных разделов сайта и имеют высокую достоверность.

**Средний-высокий** — для рейтингов качества. Экспертные оценки отражают мнение конкретных рецензентов и могут различаться для схожих наборов. Их следует рассматривать как ориентир, а не как объективную истину.

**Средний** — для национальностей. Отнесение набора к национальности выполняется рецензентами сайта на основе исторического контекста и может быть спорным для многонациональных или сложных тем.
