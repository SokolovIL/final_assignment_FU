# Шаг 24. Интерпретация результатов

В ходе проделанной работы были даны ответы на все поставленные аналитические вопросы. 

## Блок 1: Структура рынка (ABC-анализ)

### 1. Какие производители выпускают наибольшее количество наборов? Какие 20% дают 80% ассортимента?
**Факт из данных**: 18 производителей (Категория A, ~29.5% от общего числа брендов) формируют 78.9% всего ассортимента наборов (2240 из 2840).

**Объяснение смысла**: Рынок высококонцентрирован. Для базового наполнения коллекции по любой популярной теме логичнее всего ориентироваться именно на бренды Категории A, так как они обладают лучшей доступностью и предсказуемым качеством.

**Ограничение**: ABC учитывает только количество наборов, игнорируя их уникальность или цену. Категория C (длинный хвост) может содержать единственные на рынке наборы по узким историческим темам.

**Возможное действие**: При старте коллекции фокусироваться на Категории A. Переходить к Категории C только для точечного закрытия специфических исторических ниш.

### 2. Какие исторические эпохи наиболее представлены на рынке?
**Факт из данных**: Наибольшее количество наборов приходится на Новейшее время (933 набора) и Новое время (927 наборов). Древний мир и Средневековье представлены значительно меньше (433 и 380 наборов соответственно).

**Объяснение смысла**: Производители фокусируются на периодах, пользующихся наибольшим массовым спросом (Вторая мировая война, Наполеоновские войны), где проще окупить затраты на создание пресс-форм.

**Ограничение**: Абсолютное количество наборов не отражает их относительную редкость или качество.

**Возможное действие**: Коллекционерам, интересующимся Древним миром или Средневековьем, следует быть готовыми к более длительному поиску и, возможно, более высоким ценам на вторичном рынке.

### 3. Какие национальности чаще всего встречаются? Есть ли нишевые (<5 наборов)?
**Факт из данных**: В топ-5 наиболее проработанных национальностей входят German, French, British, Russian и USA. При этом существуют десятки национальностей, представленных менее чем 5 наборами.

**Объяснение смысла**: Массовый рынок ориентирован на крупные исторические державы. Нишевые национальности представляют собой "белые пятна", интересные для продвинутых коллекционеров, ищущих уникальность.

**Ограничение**: Подсчёт ведётся по упоминаниям в датасете. Один крупный набор может содержать несколько национальностей, что размывает границы.

**Возможное действие**: Использовать файл final_recommendations_rare_niches.csv для поиска уникальных наборов по редким национальностям.

## Блок 2: Динамика рынка во времени

### 4. Как менялось количество выпускаемых наборов по годам?
**Факт из данных**: С 1960-х годов наблюдается экспоненциальный рост числа релизов, с пиками в 2000-х (819 наборов) и 2010-х (854 набора). В 2020-х пока зафиксировано 321 набор (с учётом неполноты данных за последние годы).

**Объяснение смысла**: Индустрия пережила этап становления и масштабирования. Появление CAD-моделирования и глобализация рынков позволили производителям значительно увеличить частоту релизов.

**Ограничение**: Агрегация по десятилетиям сглаживает внутридекадные колебания (например, влияние экономических кризисов).

**Возможное действие**: При оценке коллекционной ценности набора учитывать, что наборы до 1980 года объективно являются более редкими.

### 5. Какие производители доминировали в разные десятилетия?
**Факт из данных**: Лидеры менялись: в 1960-х доминировал Airfix, в 1970-х — Atlantic, в 1980-х — Esci, в 1990-х — Revell, а с 2000-х годов и по настоящее время абсолютным лидером по количеству релизов является Strelets (198 наборов в 2010-х, 101 набор в 2020-х).

**Объяснение смысла**: Рынок прошёл через несколько волн консолидации. Современные лидеры вытеснили классических гигантов прошлого века за счёт более гибкой нишевой политики и адаптации к новым стандартам.

**Ограничение**: График показывает только абсолютного лидера (Топ-1) по количеству, не отражая общую долю рынка или средний рейтинг этого производителя в конкретный период.

**Возможное действие**: Наборы ушедших с рынка производителей 1960–1980-х годов могут имель большую коллекционную ценность, но сомнительное качество изготовления с точки зрения современных стандарторв. 

### 6. Есть ли слабые ниши (нет новых выпусков >15 лет)?
**Факт из данных**: Алгоритм выявления редких ниш показал, что из 2026 потенциально редких наборов 1645 (81%) являются условно избыточными, так как их исторический период уже полностью перекрывается другими наборами той же национальности. Уникально полезных для закрытия пробелов осталось всего 381 набор.

**Объяснение смысла**: Покупка избыточного набора не добавляет новой исторической ценности в коллекцию, если временной интервал и национальность уже закрыты более широким или качественным набором.

**Ограничение**: Алгоритм проверяет только математическое вложение интервалов лет. Он не оценивает разницу в качестве скульптинга или исторической достоверности фигур.

**Возможное действие**: Использовать файл final_recommendations_rare_niches.csv с фильтром is_redundant == False как чек-лист для целевой покупки.

### 7. Какие темы являются перспективными (рост за 2020–2026)?
**Факт из данных**: Наибольший темп прироста числа наборов за период 2020–2026 гг. (по сравнению с 2015–2019) показали темы: Italian (Древний мир, +129%), USA (Новейшее время, +116%) и Italian (Новейшее время, +100%). При этом темы British (Средневековье) и Confederate (Новое время) показали падение на 88% и 85% соответственно.

**Объяснение смысла**: Производители активно переориентируются на новые или ранее недостаточно охваченные ниши, в то время как классические, давно насыщенные темы теряют приоритет.

**Ограничение**: Расчёт темпа роста ограничен темами, где в базовом периоде (2015–2019) было выпущено не менее 3 наборов. Это защищает от математических аномалий, но может скрывать появление совершенно новых ниш.

**Возможное действие**: Коллекционерам, ищущим свежие новинки, стоит обратить внимание на файл promising_themes_growth.csv.

## Блок 3: Качество наборов (Aggregate_Rating)

### 8. Как менялся средний рейтинг качества наборов по годам?
**Факт из данных**: Средний рейтинг качества, низкий в середине XX века (~30–31 балл), стабилизировался на высоком уровне 41–42 балла из 50 начиная с 2000-х годов.

**Объяснение смысла**: Индустрия прошла этап становления. Современные технологии (CAD-моделирование, улучшенные пресс-формы) позволяют выпускать большие наборы с высокой детализацией, что высоко оценивается опытными коллекционерами.

**Ограничение**: Корреляция не доказывает причинность.

**Возможное действие**: При ограниченном бюджете приоритет следует отдавать наборам, выпущенным после 2000 года, как наиболее стабильным по качеству.

### 9. Какие производители имеют наивысший средний рейтинг качества?
**Факт из данных**: Категории A и B имеют схожий высокий средний рейтинг (41–42). Категория C демонстрирует несколько более низкий средний балл (38), но разрыв не является критическим.

**Объяснение смысла**: Мелкие нишевые производители (Категория C) не уступают гигантам рынка радикально. Они конкурируют не объёмом, а детализацией и уникальностью тем.

**Ограничение**: Среднее значение может скрывать разброс. У нишевых брендов одна низкая оценка сильнее влияет на среднее, чем у массовых.

**Возможное действие**: Не стоит избегать наборов производителей Категории C для закрытия специфических тем, так как их качество в среднем остаётся на конкурентоспособном уровне.

### 10. Есть ли связь между количеством фигур в наборе и рейтингом?
**Факт из данных**: Наблюдается небольшой восходящий тренд: средний рейтинг категории «Сэмплер/Диорама (51+)» составляет 42.3, что выше, чем у категории «Мини (1-15)» (40.5).

**Объяснение смысла**: Закономерность объясняется тем, что большие наборы содержат 8-12 разных фигур, что высоко оцениваются по критерию "Pose Number".

**Ограничение**: Корреляция не доказывает причинность. Высокий рейтинг может обуславливаться не размером, а брендом или конкретной исторической эпохой, преобладающей в крупных наборах.

**Возможное действие**: Для варгеймера, собирающего большую армию, покупка больших наборов (50+ фигур) статистически гарантирует более высокое качество. Для диорамиста, которому нужны 2-3 специфические фигуры, размер не имеет значения — лучше ориентироваться на рейтинг конкретного набора.

## Блок 4: Рекомендации для коллекционеров

### 11. Для конкретных исторических периодов — какие производители предлагают наборы с лучшим рейтингом?
**Факт из данных**: Для каждой эпохи отобраны топ-3 производителя с минимум 5 наборами, отсортированные по среднему рейтингу.

**Объяснение смысла**: Полученные данные дают коллекционеру конкретный, проверенный список брендов, которые являются признанными экспертами в выбранной исторической эпохе.

**Ограничение**: Усреднённый рейтинг может скрывать наличие как абсолютных хитов, так и неудачных релизов внутри одной эпохи у одного производителя.

**Возможное действие**: Использовать файл top_manufacturers_by_era.csv как основной ориентир при выборе бренда для новой коллекции.

### 12. Есть ли связь между годом выпуска и количеством фигур?
**Факт из данных**: Среднее количество фигур выросло с ~24 в 1950-х до пика в ~42 в 1980-х, после чего стабилизировалось на уровне 27–32 фигур в 2000–2020-х годах.

**Объяснение смысла**: В 1970–1980-е годы производители экспериментировали с крупными форматами. Современная стабилизация говорит о формировании отраслевого стандарта "оптимального" набора, удобного для хранения и сборки.

**Ограничение**: Агрегация по десятилетиям сглаживает возможные краткосрочные тренды внутри 10-летних периодов. Указанное среднее количество фигур не учитывает специфику наборов. Реальные цифры выше для пехих фигур и ниже для конных.  

**Возможное действие**: Ожидается, что современный стандартный набор будет содержать в среднем около 30 фигур.

## Блок 5: Адаптация RFM и XYZ под контекст коллекционирования

### 13. RFM-анализ активности производителей: Recency, Frequency, Magnitude.
**Факт из данных**: Найдено 74 "актуальных хита" (Recency ≤ 5 лет, Magnitude ≥ 40). Это пары "производитель-национальность", которые выпускают наборы недавно, часто и с высоким средним рейтингом.

**Объяснение смысла**: Эти наборы являются "горячими новинками" с доказанным качеством. Их стоит рекомендовать коллекционерам в первую очередь.

**Ограничение**: Пороговые значения (Recency ≤ 5, Magnitude ≥ 40) заданы эвристически. Изменение порогов изменит состав выборки.

**Возможное действие**: Использовать файл agg_rfm_actual_hits.csv для формирования списка приоритетных покупок.

### 14. XYZ-анализ стабильности релизов: X (стабильно), Z (спорадически).
**Факт из данных**: Из всех пар "производитель-национальность" классифицированы в категории X/Y/Z только 37% (245 пар). Оставшиеся 63% не удалось классифицировать из-за малого количества наборов (kit_count < 3). Среди классифицированных: X (стабильно) — 122, Y (волнообразно) — 109, Z (эпизодически) — 14.

**Объяснение смысла**: Коллекционер может с высокой долей уверенности рассчитывать на пополнение коллекций по темам категории X. Темы категории Z лучше закрывать через альтернативные бренды или вторичный рынок, не дожидаясь переиздания.

**Ограничение**: Это анализ стабильности предложения (факта релиза), а не спроса. Метрика regularity не учитывает объём выпусков в конкретном году. Значительная доля неклассифицированных пар — это следствие малых выборок, а не "хаотичности" производителей.

**Возможное действие**: При планировании обширной коллекции целесообразно выбирать производителей из Топ-3 по рейтингу, которые при этом имеют категорию стабильности X или Y. Если лучший производитель имеет категорию Z, набор нужно приобретать немедленно, не дожидаясь переиздания.

### Итоговое заключение по проекту
Проведённый анализ данных о рынке пластиковых исторических миниатюр позволил перейти от разрозненных каталогов к системе обоснованных рекомендаций. Было подтверждено, что рынок зрелый и поляризованный, качество наборов стабильно высоко, а ключевой проблемой коллекционеров является не отсутствие наборов, а навигация в избыточном ассортименте и редких нишах.
Разработанные метрики (адаптированные RFM и XYZ, скоринг редкости с учётом перекрытия периодов) успешно решают эту проблему, фильтруя информационный шум.
