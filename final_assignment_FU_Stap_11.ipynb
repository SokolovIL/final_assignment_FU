# Шаг 11. Обработка пропусков

На основе диагностики, проведенной на Шаге 10, и специфики датасета, будет применена стратегию, которая не искажает реальную картину рынка.

```python
import pandas as pd

df_consolidated = pd.read_csv('df_consolidated.csv')

print("=== Пропуски ДО обработки ===")
print(df_consolidated.isna().sum())

# 1. Обработка категориальных пропусков
df_consolidated['nationality'] = df_consolidated['nationality'].fillna('Не указано')
df_consolidated['manufacturer'] = df_consolidated['manufacturer'].fillna('Unknown')

# 2. Явное приведение числовых полей с пропусками к типу Int64 
# (это позволяет хранить NaN, не превращая столбец во float64 с ".0")
numeric_cols_with_nans = ['release_year', 'aggregate_rating', 'num_figures', 'decade', 'years_since_release']

for col in numeric_cols_with_nans:
    df_consolidated[col] = df_consolidated[col].astype('Int64')

print("\n=== Пропуски ПОСЛЕ обработки ===")
print(df_consolidated.isna().sum())

# 3. Проверка: сколько теперь наборов в категории "Не указано"
print("\n=== Количество наборов без указанной национальности ===")
print(f"Всего: {(df_consolidated['nationality'] == 'Не указано').sum()} строк")

# Сохраняем очищенный датасет для дальнейшего анализа и загрузки в BI
df_consolidated.to_csv('df_consolidated_clean.csv', index=False, encoding='utf-8')
print("\nОчищенный датасет сохранен как 'df_consolidated_clean.csv'")
```

## Результат шага
Пропуски обработаны способом, соответствующим предметной области.
Категориальные пропуски трансформированы в явные значения ('Не указано', 'Unknown'), что обеспечит их корректное отображение в фильтрах и сводных таблицах BI-системы.
Потери данных (удаление строк) не произошло, целостность каталога из 2840 наборов сохранена.
