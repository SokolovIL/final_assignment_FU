Будет произведено централизованое сохранение всех ключевых артефактов анализа: очищенные датасеты, агрегированные таблицы, итоговый отчёт, файл зависимостей и описание проекта.

````python
import pandas as pd
import os
from datetime import datetime

print("="*70)
print("ШАГ 28: СОХРАНЕНИЕ РЕЗУЛЬТАТОВ И ФИНАЛИЗАЦИЯ ПРОЕКТА")
print("="*70)

# =============================================================================
# 1. ПРОВЕРКА НАЛИЧИЯ ВСЕХ КЛЮЧЕВЫХ ФАЙЛОВ
# =============================================================================
print("\n--- 1. Проверка наличия всех артефактов ---")

required_files = {
    'review_periods.csv': 'Исходные данные: исторические периоды',
    'reviews_by_nationality_wide.csv': 'Исходные данные: национальности',
    'psr_enriched_data.csv': 'Исходные данные: метаданные наборов',
    'df_sets.csv': 'Очищенный датасет (уровень набора)',
    'df_consolidated.csv': 'Очищенный датасет (уровень набор-национальность)',
    'df_consolidated_clean.csv': 'Финальный очищенный датасет',
    'agg_manufacturers_abc.csv': 'ABC-анализ производителей',
    'agg_decades_dynamics.csv': 'Динамика рынка по десятилетиям',
    'agg_era_nationality_map.csv': 'Карта исторических периодов',
    'agg_figure_categories.csv': 'Связь размера набора и рейтинга',
    'agg_rfm_actual_hits.csv': 'RFM-анализ: актуальные хиты',
    'agg_manufacturers_xyz.csv': 'XYZ-анализ: стабильность присутствия',
    'final_recommendations_rare_niches.csv': 'Рекомендации: закрытие редких ниш',
    'final_recommendations_most_saturated_periods.csv': 'Рекомендации: насыщенные периоды',
    'top_manufacturers_by_nationality.csv': 'Топ производителей по национальностям',
    'top_manufacturers_by_era.csv': 'Топ производителей по эпохам',
    'top_manufacturers_by_decade.csv': 'Лидеры по десятилетиям',
    'promising_themes_growth.csv': 'Динамика тем: рост',
    'promising_themes_decline.csv': 'Динамика тем: падение',
    'heatmap_year_nationality_coverage.csv': 'Тепловая карта покрытия (полная)',
    'README.md': 'Описание проекта',
    'FINAL_REPORT.md': 'Итоговый аналитический отчёт',
    'requirements.txt': 'Зависимости Python'
}

all_present = True
for file, description in required_files.items():
    if os.path.exists(file):
        size_kb = os.path.getsize(file) / 1024
        print(f"✅ {file:50s} ({size_kb:6.1f} KB) — {description}")
    else:
        print(f"❌ {file:50s} — ОТСУТСТВУЕТ!")
        all_present = False

if all_present:
    print("\n✅ ВСЕ АРТЕФАКТЫ НА МЕСТЕ. Проект готов к сдаче.")
else:
    print("\n⚠️ ВНИМАНИЕ: Некоторые файлы отсутствуют. Проверьте предыдущие шаги.")

# =============================================================================
# 2. СОЗДАНИЕ ФАЙЛА ТРЕБОВАНИЙ (requirements.txt)
# =============================================================================
print("\n--- 2. Создание requirements.txt ---")

with open('requirements.txt', 'w', encoding='utf-8') as f:
    f.write("# Зависимости Python для воспроизведения анализа\n")
    f.write("# Установка: pip install -r requirements.txt\n\n")
    f.write("pandas>=2.0.0\n")
    f.write("numpy>=1.24.0\n")
    f.write("matplotlib>=3.7.0\n")
    f.write("seaborn>=0.12.0\n")
    f.write("jupyter>=1.0.0\n")
    f.write("openpyxl>=3.1.0\n")

print("✅ Создан файл 'requirements.txt' с зависимостями Python.")

# =============================================================================
# 3. СОЗДАНИЕ ФАЙЛА С ИНСТРУКЦИЕЙ ПО ВОСПРОИЗВЕДЕНИЮ
# =============================================================================
print("\n--- 3. Создание REPRODUCIBILITY.md ---")

with open('REPRODUCIBILITY.md', 'w', encoding='utf-8') as f:
    f.write("# Инструкция по воспроизведению анализа\n\n")
    f.write("## Требования\n")
    f.write("- Python 3.9 или выше\n")
    f.write("- Jupyter Notebook или JupyterLab\n")
    f.write("- Интернет-соединение (для публикации BI-дашборда)\n\n")
    f.write("## Установка окружения\n\n")
    f.write("```bash\n")
    f.write("# Создайте виртуальное окружение\n")
    f.write("python -m venv venv\n")
    f.write("source venv/bin/activate  # Linux/Mac\n")
    f.write("# или\n")
    f.write("venv\\Scripts\\activate  # Windows\n\n")
    f.write("# Установите зависимости\n")
    f.write("pip install -r requirements.txt\n")
    f.write("```\n\n")
    f.write("## Порядок запуска ноутбуков\n\n")
    f.write("Запускайте ноутбуки строго в указанном порядке:\n\n")
    for i in range(29):
        f.write(f"{i+1}. `final_assignment_FU_Stap_{i}.ipynb`\n")
    f.write("\n## Воспроизведение BI-дашборда\n\n")
    f.write("1. Загрузите CSV-файлы из папки проекта в BI-платформу (Yandex DataLens / Power BI).\n")
    f.write("2. Постройте дашборд с фильтрами по эпохам, национальностям и категориям ABC/XYZ.\n")
    f.write("3. Опубликуйте дашборд и добавьте ссылку в `README.md`.\n\n")
    f.write("## Контакт\n\n")
    f.write("По вопросам воспроизведения обращайтесь к автору проекта.\n")

print("✅ Создан файл 'REPRODUCIBILITY.md' с инструкцией по воспроизведению.")

# =============================================================================
# 4. ИТОГОВАЯ СТАТИСТИКА ПРОЕКТА
# =============================================================================
print("\n--- 4. Итоговая статистика проекта ---")

notebooks = len([f for f in os.listdir('.') if f.endswith('.ipynb')])
csv_files = len([f for f in os.listdir('.') if f.endswith('.csv')])
md_files = len([f for f in os.listdir('.') if f.endswith('.md')])

print(f"📊 Статистика проекта:")
print(f"   • Jupyter Notebooks: {notebooks}")
print(f"   • CSV-файлы: {csv_files}")
print(f"   • Markdown-документы: {md_files}")
print(f"   • Дата генерации: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")

print("\n" + "="*70)
print("🎉 ШАГ 28 ЗАВЕРШЁН УСПЕШНО!")
print("Все артефакты сохранены, проект готов к финальной сдаче.")
print("="*70)
````
